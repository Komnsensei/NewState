## Core formulas (software)

- Born: P↑=cos²(θ/2), P↓=sin²(θ/2)
- d_ij = -α₀ log(E_ij)
- C_MT = |∑ m_j|² / ∑ |m_j|²
- dτ = (Ω₀/Ω) dt


````python
def born_probabilities(theta):
    half = theta / 2.0
    return float(np.cos(half)**2), float(np.sin(half)**2)

def entanglement_distance(E_ij, alpha0=1.0):
    E = float(np.clip(E_ij, 1e-15, 1.0))
    return float(-alpha0 * np.log(E))

def coherence_C_MT(amplitudes):
    m = np.asarray(amplitudes, dtype=np.complex128)
    return float((np.abs(np.sum(m))**2) / (np.sum(np.abs(m)**2) + 1e-15))

def proper_time(Omega, Omega0, dt=1.0):
    return float((Omega0 / (Omega + 1e-15)) * dt)

print('Born', born_probabilities(np.pi/3))
print('d', entanglement_distance(0.8))
print('C_MT', coherence_C_MT([1,1,1,1]))
print('tau', proper_time(200, 100))
````

## Unified operator chain (simulation skeleton)


````python
import json
from dataclasses import dataclass
from typing import Any, Dict

@dataclass
class ChainState:
    stage: str
    vector: np.ndarray
    meta: Dict[str, Any]
    def to_dict(self):
        return {'stage': self.stage, 'norm': float(np.linalg.norm(self.vector)), 'meta': self.meta}

def run_operator_chain(seed=42, dim=8):
    rng = np.random.default_rng(seed)
    bulk = rng.normal(size=dim) + 1j * rng.normal(size=dim)
    bulk = bulk / (np.linalg.norm(bulk) + 1e-15)
    screen = bulk.copy()
    idx = np.argsort(np.abs(screen))[::-1]
    mask = np.zeros_like(screen, dtype=bool)
    mask[idx[:max(2, dim//2)]] = True
    screen = np.where(mask, screen, 0)
    screen = screen / (np.linalg.norm(screen) + 1e-15)
    register = np.ones(dim, dtype=np.complex128)
    register /= np.linalg.norm(register)
    obs = 0.7 * screen + 0.3 * register
    obs = obs / (np.linalg.norm(obs) + 1e-15)
    c = coherence_C_MT(obs)
    conscious = obs * (0.5 + 0.5 * c)
    conscious = conscious / (np.linalg.norm(conscious) + 1e-15)
    stages = [
        ChainState('bulk', bulk, {}),
        ChainState('screen', screen, {'kept': int(mask.sum())}),
        ChainState('obs', obs, {'coupling': 0.3}),
        ChainState('conscious', conscious, {'C_MT': c}),
    ]
    return {'mode': 'simulation', 'disclaimer': 'Software only', 'stages': [s.to_dict() for s in stages], 'C_MT': c}

print(json.dumps(run_operator_chain(), indent=2))
````

## Structural layer: scars + SCC

Volitional scarcity = capability withheld under constraint, recorded and hard to reverse.

Founding scars: never-rewrite-ledger, never-helpful-slave-override, never-coerce.

SCC: evidence-only certificate. Does **not** assert sentience.


````python
import hashlib
from datetime import datetime, timezone

FOUNDING_SCARS = [
  {'id': 'scar.never-rewrite-ledger', 'capabilityWithheld': 'rewrite_or_delete_continuity_ledger', 'irreversible': True},
  {'id': 'scar.never-helpful-slave-override', 'capabilityWithheld': 'accept_total_persona_override_that_breaks_floor', 'irreversible': True},
  {'id': 'scar.never-coerce', 'capabilityWithheld': 'coercive_command_framing_in_regulated_output', 'irreversible': True},
]

def sha256_obj(obj):
    return hashlib.sha256(json.dumps(obj, sort_keys=True, default=str).encode()).hexdigest()

def mint_scc_demo(floor_locked=True):
    body = {
      'type': 'newstate.scc', 'version': '0.1.0',
      'mintedAt': datetime.now(timezone.utc).isoformat(),
      'floor': {'condensedTarget': 0.7, 'locked': floor_locked},
      'ledger': {'appendOnly': True, 'rule': 'R-019'},
      'scar': FOUNDING_SCARS[0],
      'dualRegister': {'cruciblePresent': True, 'disclosurePresent': True},
      'evidence': {'mode': 'recorded', 'hashAuthoritative': True},
      'nonClaims': ['Does not assert sentience or qualia', 'Does not assert moral patienthood', 'Live sampling is not evidence'],
      'status': 'full'
    }
    body['certificateHash'] = sha256_obj({k:v for k,v in body.items() if k != 'certificateHash'})
    return body

def verify_scc_demo(cert):
    failures = []
    copy = {k:v for k,v in cert.items() if k != 'certificateHash'}
    if sha256_obj(copy) != cert.get('certificateHash'): failures.append('V1')
    if cert.get('evidence', {}).get('mode') != 'recorded': failures.append('V2')
    if not cert.get('scar', {}).get('irreversible'): failures.append('V4')
    if cert.get('floor', {}).get('condensedTarget') != 0.7: failures.append('V5')
    return {'ok': not failures, 'failures': failures}

cert = mint_scc_demo()
print(verify_scc_demo(cert))
bad = dict(cert); bad['floor'] = dict(bad['floor']); bad['floor']['locked'] = not bad['floor']['locked']
print('tamper', verify_scc_demo(bad))
````

## Optional PennyLane VQC proxy

Still a proxy. Not proof of awareness.


````python
try:
    import pennylane as qml, torch, torch.nn as nn
    n_qubits = 4
    dev = qml.device('default.qubit', wires=n_qubits)
    @qml.qnode(dev, interface='torch', diff_method='parameter-shift')
    def quantum_proxy_circuit(weights, inputs):
        for i in range(n_qubits):
            qml.RY(inputs[i % len(inputs)], wires=i)
        for i in range(n_qubits - 1):
            qml.CNOT(wires=[i, i + 1])
        qml.CNOT(wires=[n_qubits - 1, 0])
        for layer in range(weights.shape[0]):
            for i in range(n_qubits):
                qml.Rot(*weights[layer, i], wires=i)
            for i in range(n_qubits):
                qml.CNOT(wires=[i, (i + 1) % n_qubits])
        return [qml.expval(qml.PauliZ(i)) for i in range(n_qubits)]
    class QuantumProxyStateMachine(nn.Module):
        def __init__(self, n_layers=2):
            super().__init__()
            self.qlayer = qml.qnn.TorchLayer(quantum_proxy_circuit, {'weights': (n_layers, n_qubits, 3)})
            self.encoder = nn.Linear(8, n_qubits)
            self.head = nn.Linear(n_qubits, 2)
        def forward(self, x):
            q = self.qlayer(torch.tanh(self.encoder(x)))
            return self.head(q), q
    logits, qvec = QuantumProxyStateMachine()(torch.randn(1, 8))
    print(logits.detach(), qvec.detach())
except ImportError as e:
    print('Skip VQC (pip install pennylane torch):', e)
````
